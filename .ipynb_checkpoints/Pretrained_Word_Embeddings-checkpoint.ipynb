## Retraining on your own corpus

```python
wget.download('https://raw.githubusercontent.com/mxw/grmr/master/src/finaltests/bible.txt')
```

```python
with open('bible.txt', 'r') as f:
  bible= f.read()
```

```python
print(bible[:1000])
```

```python
# Cleaning the text
import re
cleaned= re.sub('[0-9][0-9]?:[0-9][0-9]? ',"", bible)
```

```python
print(cleaned[:1000])
```

```python
import nltk
nltk.download('all',quiet= True)
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
sentences= sent_tokenize(cleaned)
```

```python
processed_data= []
for sent in sentences:
  processed_data.append(word_tokenize(sent))
```

```python
processed_data[:5]
```

*After the data is processed, let build our own Word2Vec*

```python
# Training on first 10 sentences just for the sake visualising word vectors 
from gensim.models import Word2Vec
Model= Word2Vec(processed_data[:10], size=300, sg=0) # Continuous Bag of Words (CBOW) Model, sg=0
```

```python
Model['God']
```

```python
len(Model['God'])
```

```python
# fit a 2d PCA model to the vectors
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
X = Model[Model.wv.vocab]
pca = PCA(n_components=2)
result = pca.fit_transform(X)
# create a scatter plot of the projection
plt.scatter(result[:, 0], result[:, 1])
words = list(Model.wv.vocab)
for i, word in enumerate(words):
    plt.annotate(word, xy=(result[i, 0], result[i, 1]))
plt.show()
```

```python
Model.similar_by_word('God')
```

**To retrain on google word2Vec we need to use intersect_word2vec_format**

*Improving your vectors using Google Word2Vec.*

```python
Model.intersect_word2vec_format('GoogleNews-vectors-negative300.bin.gz',
                                lockf=1.0,
                                binary=True)
```

```python
# Training it
Model.train(processed_data[:10], total_examples=Model.corpus_count, epochs=10)
```

```python
Model.similar_by_word('God')
```

```python
# fit a 2d PCA model to the vectors
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
X = Model[Model.wv.vocab]
pca = PCA(n_components=2)
result = pca.fit_transform(X)
# create a scatter plot of the projection
plt.scatter(result[:, 0], result[:, 1])
words = list(Model.wv.vocab)
for i, word in enumerate(words):
    plt.annotate(word, xy=(result[i, 0], result[i, 1]))
plt.show()
```

## Stanford's Glove
[GloVe: Global Vectors for Word Representation](https://nlp.stanford.edu/projects/glove/)

*GloVe is an unsupervised learning algorithm for obtaining vector representations for words. Training is performed on aggregated global word-word co-occurrence statistics from a corpus, and the resulting representations showcase interesting linear substructures of the word vector space.*

*You can download the smallest GloVe pre-trained model from the GloVe website. It an 822 Megabyte zip file with 4 different models (50, 100, 200 and 300-dimensional vectors) trained on Wikipedia data with 6 billion tokens and a 400,000 word vocabulary.*

```python
wget.download('http://nlp.stanford.edu/data/glove.6B.zip')
```

```python
import zipfile
zip= zipfile.ZipFile('glove.6B.zip')
zip.extractall()
```

```python
from gensim.test.utils import datapath, get_tmpfile
from gensim.models import KeyedVectors
from gensim.scripts.glove2word2vec import glove2word2vec
glove_input_file = 'glove.6B.300d.txt'
word2vec_output_file = 'glove.6B.word2vec'
_ = glove2word2vec(glove_input_file, word2vec_output_file)
model = KeyedVectors.load_word2vec_format(word2vec_output_file)
```

# Facebooks's fastText

[fastText](https://fasttext.cc/) *is a library for learning of word embeddings and text classification created by Facebook's AI Research (FAIR) lab. The model allows to create an unsupervised learning or supervised learning algorithm for obtaining vector representations for words. Facebook makes available pretrained models for 294 languages.fastText uses a neural network for word embedding.*

> Source: Wikipedia

**Installation:**

> !pip install fasttext


*The FastText model was first introduced by Facebook in 2016 as an extension and supposedly improvement of the vanilla Word2Vec model. If you want to get in-depth knowledge of FastText, go thorugh this paper titled [Enriching Word Vectors with Subword Information’ by Mikolov et al.](https://arxiv.org/pdf/1607.04606.pdf)*

The framework is open-sourced by Facebook on GitHub and claims to have the following.
- Recent state-of-the-art English word vectors.
- Word vectors for 157 languages trained on Wikipedia and Crawl.
- Models for language identification and various supervised tasks.

## How is it different from Word2Vec??

*Basically Word2Vec ignores the morphology of words, by assigning a distinct vector to each. Now this is limitation, especially with large vocabularies and many rare words. FastText embeddings is a new approch based on the skipgram model, where each word is represented as a bag of character n-grams. A vector representation is associated to each character n-gram; words being represented as sum of these representations.*

- It learns word embeddings within the word
- Each token/word will be expressed as sum and average of it's n-gram componenets
> `kingdom= ['k', 'ki', 'kin', 'king', 'kingd', 'kingdo', 'kingdom']`
- It can generate word embeddings for Out-of-Vocabulary words but Word2vec and glove cannot.
- It will be more accuarte
- As words vector contains information about its subwords, So initially it will take up more memory and the computation/ training time with increase drastically.

# ELMO - Word Embeddings on Steroids

- The meaning of a word is context-dependent; their embeddings should also take context into account

- Embeddings from Language Models (ELMos) use language models to obtain embeddings for individual words while taking the entire sentence or paragraph into account.

- Concretely, ELMos use a pre-trained, multi-layer, bi-directional, LSTM-based language model and extract the hidden state of each layer for the input sequence of words. Then, they compute a weighted sum of those hidden states to obtain an embedding for each word.

- The weight of each hidden state is task-dependent and is learned.

- ELMo improves the performance of models across a wide range of tasks, spanning from question answering and sentiment analysis to named entity recognition.

*To get pretrained word embeddings from ELMO we will use AllenNLP Library.*

> *pip install allennlp*

```python
!pip -q install allennlp
```

```python
from allennlp.modules.elmo import Elmo, batch_to_ids
```

```python
options_file = "https://allennlp.s3.amazonaws.com/models/elmo/2x4096_512_2048cnn_2xhighway/elmo_2x4096_512_2048cnn_2xhighway_options.json"
weight_file = "https://allennlp.s3.amazonaws.com/models/elmo/2x4096_512_2048cnn_2xhighway/elmo_2x4096_512_2048cnn_2xhighway_weights.hdf5"
```

```python
# Compute two different representation for each token.
# Each representation is a linear weighted combination for the
# 3 layers in ELMo (i.e., charcnn, the outputs of the two BiLSTM))
elmo = Elmo(options_file, weight_file, 2, dropout=0)
```

```python
# use batch_to_ids to convert sentences to character ids
sentences = [['This', 'is', 'my', 'first', 'sentence', '.'], ['This', 'is', 'another', '.']]
character_ids = batch_to_ids(sentences)
```

```python
embeddings = elmo(character_ids)
```

```python
embeddings
```

*If you are not training a pytorch model, and just want numpy arrays as output then use*
> allennlp.commands.elmo.ElmoEmbedder

```python
from allennlp.commands.elmo import ElmoEmbedder
elmo = ElmoEmbedder()
tokens = ["I", "ate", "an", "apple", "for", "breakfast"]
vectors = elmo.embed_sentence(tokens)
```
