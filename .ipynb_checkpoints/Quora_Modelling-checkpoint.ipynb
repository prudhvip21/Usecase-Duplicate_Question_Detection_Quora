```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
pd.set_option('display.max_columns', 100) # Show all columns when looking at dataframe
import warnings
warnings.filterwarnings('ignore') # remove warnings from the notebook
from IPython.display import Math, HTML
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
import re
import string
from nltk import ngrams
```

```python
train= pd.read_csv('final_df.csv')
```

# Model Building

**Defining our dependent and independent variable.**

```python
X= train[['stem_uni_match', 'uni_match', 'wm_distance' , 'bi_match', 'cos_sim']]
y= train.is_duplicate
```

```python
X= np.nan_to_num(X)
```

*We will work on sample of Data Set, because training on half a million Data points is computationaly expensive.*

*To remove imbalance in our dataset we are going to take 20000 datapoints from duplicate class and 20000 from non-duplicate Class.*

## UnderSampling

```python
from imblearn.under_sampling import RandomUnderSampler
sampler = RandomUnderSampler(ratio={1: 20000, 0: 20000})
X_s, y_s= sampler.fit_sample(X, y)
```

```python
X_s.shape, y_s.shape
```

## Test Train Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_s, y_s, test_size=0.2, random_state=42)
```

## Logistic Regression

```python
## Fitting Model to the Training Set
from sklearn.linear_model import LogisticRegression
lg_clf = LogisticRegression(random_state = 0)
lg_clf.fit(X_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, precision_score, recall_score
from sklearn.metrics import roc_curve, auc
from sklearn.model_selection import cross_val_score
results = pd.DataFrame( columns = ['Model', 'Accuracy', 'Precision', 'Recall', 'F1 Score', 
                          'GridSearch Accuracy Training Set (5 Folds)'])
```

```python
# Classificaion Evaluation Template
def classification_results(classifier):
    # Predicting Test Set
    y_pred = classifier.predict(X_test)
    # Evaluating Results
    cm = confusion_matrix(y_test, y_pred)
    acc= accuracy_score(y_test, y_pred)
    precision= precision_score(y_test, y_pred) # tp / (tp + fp)
    recall= recall_score(y_test, y_pred) # tp / (tp + fn)
    f1= f1_score(y_test, y_pred)
    print(f"Accuracy : {(acc* 100):.4f} %\n\n\n")
    # Printing the Confusion Matrix with test Accuracy
    df_cm = pd.DataFrame(cm, index = (0, 1), columns = (0, 1))
    plt.figure(figsize = (7,5))
    sns.set(font_scale=1.4)
    sns.heatmap(df_cm, annot=True, fmt='g', xticklabels=['Non Duplicate', 'Duplicate'], yticklabels=['Non Duplicate', 'Duplicate'])
    plt.xticks(rotation=45)
    plt.yticks(rotation='horizontal')
    plt.title("CONFUSION MATRIX", fontsize= 18)
    plt.show();
    print("\n\n\n")
    accuracies = cross_val_score(estimator = classifier, X = X_train, y = y_train, cv = 5)
    # plotting the ROC Curve
    y_predict_probabilities = classifier.predict_proba(X_test)[:,1]
    fpr, tpr, _ = roc_curve(y_test, y_predict_probabilities)
    roc_auc = auc(fpr, tpr)
    plt.figure(figsize=(8,6))
    plt.plot(fpr, tpr, color='darkorange',
         lw=2, label='ROC curve (area = %0.2f)' % roc_auc)
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend(loc="lower right")
    plt.show();
    model_name= re.findall("(.*?)\(", str(classifier.fit(X_train, y_train)))[0]
    model_results = pd.DataFrame([[model_name, acc, precision, recall, f1, 
                         f"{accuracies.mean():.3f} ( +/- {(accuracies.std() * 2):.3f})"]],
               columns = ['Model', 'Accuracy', 'Precision', 'Recall', 'F1 Score', 
                          'GridSearch Accuracy Training Set (5 Folds)'])
    global results
    results = results.append(model_results, ignore_index = True)
```

```python
# Logistic Regression Results
classification_results(lg_clf)
```

```python
results
```

## SVM (Linear)

```python
from sklearn.svm import SVC
svm_clf = SVC(random_state = 0, kernel = 'linear', probability=True)
svm_clf.fit(X_train, y_train)
```

```python
classification_results(svm_clf)
```

```python
results
```

## K Nearest Neighbors Classifier

```python
from sklearn.neighbors import KNeighborsClassifier
knn_clf = KNeighborsClassifier(n_neighbors=13)
knn_clf.fit(X_train, y_train) 
```

```python
classification_results(knn_clf)
```

```python
results
```

## Gaussian Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
gb_clf = GaussianNB()
gb_clf.fit(X_train, y_train)
```

```python
classification_results(gb_clf)
```

```python
results
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
rf_clf = RandomForestClassifier(random_state = 0, n_estimators = 100, criterion = 'entropy')
rf_clf.fit(X_train, y_train)
```

```python
classification_results(rf_clf)
```

```python
results
```

## Model Selection

*We will go with Randon Forest Model as it has decent acuuracy and diffrence between Precision and Recall is not much high.*

```python
y_pred= rf_clf.predict(X_test)
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred, target_names=['Non-Duplicate', 'Duplicate']))
```
